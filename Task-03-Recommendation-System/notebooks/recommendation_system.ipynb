
# Netflix Content-Based Recommendation System

## 1. Project Overview

The objective of this project is to develop a content-based recommendation system using the Netflix titles dataset.

The system analyzes movie and TV show information such as title, genre, description, cast, director, and other relevant features to identify similarities between different titles.

Using Natural Language Processing (NLP) techniques and Machine Learning, the system will recommend Netflix titles based on the content characteristics of a selected movie or TV show.

## 2. Project Objectives

- Extract relevant content features from the Netflix dataset.
- Perform text preprocessing and cleaning.
- Convert textual information into numerical representations using TF-IDF.
- Calculate similarity between Netflix titles using Cosine Similarity.
- Generate recommendations for selected titles.
- Evaluate the quality and relevance of recommendations.

## 4. Recommendation Approach

This project uses Content-Based Filtering.

Content-Based Filtering recommends movies or TV shows based on their similarities in content and characteristics rather than relying on ratings or preferences of other users.

## 5. Expected Outcome

A recommendation model that accepts a Netflix movie or TV show title as input and returns a list of similar titles based on their content features.

```python

# Select relevant features for recommendation

features = [
    'title',
    'type',
    'director',
    'country',
    'rating',
    'listed_in'
]

# Create a new DataFrame containing selected features

recommendation_df = df[features].copy()

# Display the first five records

recommendation_df.head()
```

```python

# Check the structure of selected features

print("Selected Features:", recommendation_df.columns.tolist())

print("\nDataset Shape:", recommendation_df.shape)

print("\nMissing Values:")
print(recommendation_df.isnull().sum())

print("\nData Types:")
print(recommendation_df.dtypes)
```

```python

# Import text preprocessing library

import re

print("Text preprocessing library imported successfully!")
```

```python

# Define a function for text preprocessing

def clean_text(text):

    # Handle missing values
    text = str(text)

    # Convert text to lowercase
    text = text.lower()

    # Remove punctuation and special characters
    text = re.sub(r'[^a-zA-Z0-9\s]', ' ', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    return text


# Select content features for preprocessing
text_features = ['type', 'director', 'country', 'rating', 'listed_in']

# Apply text preprocessing to each feature
for column in text_features:
    recommendation_df[column + '_clean'] = recommendation_df[column].apply(clean_text)


# Display original and cleaned features
recommendation_df[['title', 'listed_in', 'listed_in_clean',
                   'director', 'director_clean']].head()
```

```python

# Combine all cleaned content features into one column

recommendation_df['combined_features'] = (
    recommendation_df['type_clean'] + ' ' +
    recommendation_df['director_clean'] + ' ' +
    recommendation_df['country_clean'] + ' ' +
    recommendation_df['rating_clean'] + ' ' +
    recommendation_df['listed_in_clean']
)

# Display title and combined features

recommendation_df[['title', 'combined_features']].head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Create the TF-IDF vectorizer
tfidf = TfidfVectorizer(stop_words='english')

# Convert combined text features into numerical vectors
tfidf_matrix = tfidf.fit_transform(
    recommendation_df['combined_features']
)

# Display the results
print("TF-IDF matrix shape:", tfidf_matrix.shape)
print("Number of unique words:", len(tfidf.get_feature_names_out()))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Select the first Netflix title
selected_index = 0

# Calculate similarity with all other titles
similarity_scores = cosine_similarity(
    tfidf_matrix[selected_index],
    tfidf_matrix
).flatten()

# Display the first 10 similarity scores
print("Selected Title:", recommendation_df.iloc[selected_index]['title'])

print("\nSimilarity Scores:")
print(similarity_scores[:10])
```

```python
def recommend_titles(title, num_recommendations=5):

    # Find the index of the selected title
    title_matches = recommendation_df[
        recommendation_df['title'].str.lower() == title.lower()
    ]

    # Check whether the title exists
    if title_matches.empty:
        print("Title not found in the dataset!")
        return

    # Get the index of the selected title
    selected_index = title_matches.index[0]

    # Calculate similarity with all titles
    similarity_scores = cosine_similarity(
        tfidf_matrix[selected_index],
        tfidf_matrix
    ).flatten()

    # Sort titles by similarity (highest first)
    similar_indices = similarity_scores.argsort()[::-1]

    # Remove the selected title itself
    similar_indices = [
        i for i in similar_indices if i != selected_index
    ]

    # Select the top recommendations
    top_indices = similar_indices[:num_recommendations]

    # Create recommendation results
    recommendations = recommendation_df.iloc[top_indices][
        ['title', 'type', 'listed_in']
    ].copy()

    # Add similarity scores
    recommendations['similarity_score'] = similarity_scores[top_indices]

    return recommendations.reset_index(drop=True)
```

```python
recommend_titles("Dick Johnson Is Dead", 5)
```

```python
# Get the original title's categories
selected_title = "Dick Johnson Is Dead"

selected_categories = set(
    recommendation_df.loc[
        recommendation_df['title'] == selected_title,
        'listed_in'
    ].iloc[0].split(', ')
)

# Generate recommendations
results = recommend_titles(selected_title, 5)

# Calculate category overlap
for _, row in results.iterrows():

    recommended_categories = set(row['listed_in'].split(', '))

    common_categories = selected_categories.intersection(
        recommended_categories
    )

    print("Recommended:", row['title'])
    print("Common Categories:", common_categories)
    print("Number of Common Categories:", len(common_categories))
    print("-" * 50)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Create a category-focused TF-IDF vectorizer
category_tfidf = TfidfVectorizer(stop_words='english')

# Convert categories into numerical vectors
category_matrix = category_tfidf.fit_transform(
    recommendation_df['listed_in_clean']
)

print("Category TF-IDF matrix shape:", category_matrix.shape)
print("Unique category terms:",
      len(category_tfidf.get_feature_names_out()))
```

```python
def recommend_by_category(title, num_recommendations=5):

    # Find the selected title
    title_matches = recommendation_df[
        recommendation_df['title'].str.lower() == title.lower()
    ]

    if title_matches.empty:
        print("Title not found in the dataset!")
        return

    # Get its index
    selected_index = title_matches.index[0]

    # Calculate category similarity
    scores = cosine_similarity(
        category_matrix[selected_index],
        category_matrix
    ).flatten()

    # Sort scores from highest to lowest
    similar_indices = scores.argsort()[::-1]

    # Exclude the selected title
    similar_indices = [
        i for i in similar_indices if i != selected_index
    ]

    # Select top recommendations
    top_indices = similar_indices[:num_recommendations]

    # Prepare results
    results = recommendation_df.iloc[top_indices][
        ['title', 'type', 'listed_in']
    ].copy()

    results['similarity_score'] = scores[top_indices]

    return results.reset_index(drop=True)
```

```python
recommend_by_category("Dick Johnson Is Dead", 5)
```

```python
# Select a test title
selected_title = "Dick Johnson Is Dead"

# Get the selected title's categories
selected_categories = set(
    recommendation_df.loc[
        recommendation_df['title'] == selected_title,
        'listed_in'
    ].iloc[0].split(', ')
)

# Generate category-based recommendations
results = recommend_by_category(selected_title, 5)

# Count relevant recommendations
relevant_count = 0

for _, row in results.iterrows():

    recommended_categories = set(row['listed_in'].split(', '))

    common_categories = selected_categories.intersection(
        recommended_categories
    )

    if common_categories:
        relevant_count += 1

# Calculate category-match percentage
match_percentage = (relevant_count / len(results)) * 100

print("Selected Title:", selected_title)
print("Total Recommendations:", len(results))
print("Category Matches:", relevant_count)
print("Category Match Percentage:", match_percentage, "%")
```

```python
import os

# Create the output folder if it does not exist
os.makedirs('../outputs', exist_ok=True)

# Generate recommendations
final_recommendations = recommend_by_category(
    "Dick Johnson Is Dead", 5
)

# Save recommendations as CSV
final_recommendations.to_csv(
    '../outputs/recommendations.csv',
    index=False
)

print("Recommendations saved successfully!")
```
