```python
#step - 1
import pandas as pd
import numpy as np

#loading the dataset

df = pd.read_csv(r"D:\aiml and datascience\Auspify-Data-Science-Internship\Task-01-Data-Cleaning\data\raw\Dataset.csv")

#first five record

df.head()
```

```python
#step-2

# Check the number of rows and columns

print("Dataset Shape:", df.shape)

# Display all column names

print("\nColumn Names:")
print(df.columns.tolist())

# Display dataset information

print("\nDataset Information:")
df.info()
```

```python
#step-1
# Check the number of rows and columns

print("Dataset Shape:", df.shape)

# Display all column names

print("\nColumn Names:")
print(df.columns.tolist())

# Display dataset information

print("\nDataset Information:")
df.info()
```

```python

#step-2
# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Check total missing values
print("\nTotal Missing Values:", df.isnull().sum().sum())

# Check duplicate records
print("\nDuplicate Records:", df.duplicated().sum())

# Check empty strings
print("\nEmpty Strings:")
print((df.astype(str).apply(lambda col: col.str.strip().eq("")).sum()))
```

```python
#step - 3 : handling null values and inconsistent format
# Check common placeholder values

placeholders = ["Unknown", "Not Given", "N/A", "NA", "null", "None"]

for column in df.columns:
    count = df[column].astype(str).str.strip().isin(placeholders).sum()

    if count > 0:
        print(f"{column}: {count} placeholder values")
```

```python

#step - 3 : handling null values and inconsistent format

# Find the placeholder values in each affected column

for column in ["title", "director", "country"]:

    print(f"\nColumn: {column}")

    print(
        df.loc[
            df[column].astype(str).str.strip().isin(placeholders),
            column
        ].value_counts()
    )

# Inspect the record with the missing title

print("\nRecord with placeholder title:")

print(
    df.loc[
        df["title"].astype(str).str.strip().isin(placeholders)
    ]
)
```

```python


'''The Unknown title is actually a real movie. It is the 2011 movie directed by Jaume Collet-Serra. Therefore, we must preserve it, not treat it as missing data.

For the other two columns, Not Given represents unavailable information. We'll standardize it to Not Available.'''

# Replace placeholder values in director and country

df["director"] = df["director"].replace("Not Given", "Not Available")

df["country"] = df["country"].replace("Not Given", "Not Available")

print("Placeholder replacement completed.")
```

```python
# Verify director values

print("Director - Not Given:", (df["director"] == "Not Given").sum())

print("Director - Not Available:", (df["director"] == "Not Available").sum())


# Verify country values

print("\nCountry - Not Given:", (df["country"] == "Not Given").sum())

print("Country - Not Available:", (df["country"] == "Not Available").sum())


# Verify the movie title is preserved

print("\nUnknown movie title:")
print(df[df["title"] == "Unknown"])
```

```python
# Check the current data type

print("Before conversion:")
print(df["date_added"].dtype)


# Convert date_added into datetime format

df["date_added"] = pd.to_datetime(
    df["date_added"],
    errors="coerce"
)


# Check the data type after conversion

print("\nAfter conversion:")
print(df["date_added"].dtype)


# Display sample dates

print("\nSample dates:")
print(df["date_added"].head(10))


# Check missing dates created during conversion

print("\nMissing dates after conversion:")
print(df["date_added"].isnull().sum())
```

```python
# Check unique duration formats

print("Unique duration units:")

print(
    df["duration"]
    .str.extract(r"([A-Za-z]+)$")[0]
    .value_counts()
)

# Check missing duration values

print("\nMissing duration values:")
print(df["duration"].isnull().sum())
```

```python
# Extract numerical duration

df["duration_value"] = df["duration"].str.extract(r"(\d+)").astype(int)


# Extract duration unit

df["duration_unit"] = df["duration"].str.extract(r"([A-Za-z]+)$")


# Standardize Season and Seasons

df["duration_unit"] = df["duration_unit"].replace({
    "Seasons": "Season"
})


# Display the transformed data

print(df[["type", "duration", "duration_value", "duration_unit"]].head(10))
```

```python
# Check duration unit counts

print("Duration Unit Counts:")
print(df["duration_unit"].value_counts())


# Check whether any duration values are missing

print("\nMissing Duration Values:")
print(df["duration_value"].isnull().sum())


# Check whether Movie records have minutes

print("\nMovie Duration Units:")
print(df.loc[df["type"] == "Movie", "duration_unit"].value_counts())


# Check whether TV Shows have seasons

print("\nTV Show Duration Units:")
print(df.loc[df["type"] == "TV Show", "duration_unit"].value_counts())


# Check final dataset dimensions

print("\nFinal Dataset Shape:", df.shape)
```

```python
# Step 4: Transform categorical and date-related columns

# Inspect unique values in categorical columns

categorical_columns = ["type", "rating"]

for column in categorical_columns:
    print(f"\nUnique values in {column}:")
    print(df[column].value_counts(dropna=False))
```

```python
# Remove unnecessary spaces from text columns

text_columns = df.select_dtypes(include="object").columns

for column in text_columns:
    df[column] = df[column].str.strip()

print("Text formatting standardized successfully.")
```

```python
# Extract year from date_added
df["date_added_year"] = df["date_added"].dt.year

# Extract month number
df["date_added_month"] = df["date_added"].dt.month

# Extract month name
df["date_added_month_name"] = df["date_added"].dt.month_name()

# Display the transformed columns
print(df[[
    "date_added",
    "date_added_year",
    "date_added_month",
    "date_added_month_name"
]].head(10))
```
